```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style="whitegrid", palette="YlGnBu")
df=pd.read_csv('train_cleaned_f2.csv')
df.head()
```

```python
df.info()
df.describe()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Income',palette='YlGnBu')
plt.title('Distribution of Income (Target Variable)',fontweight='bold',pad=10)

plt.xticks(ticks=[0, 1], labels=[r'>=50K', r'>50K'])
plt.legend(title='Income Level', labels=['=>50K', '>50K'])
plt.show()
```

```python

plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='age', kde=True, color='skyblue')
plt.title('Age Distribution of Individuals',fontweight='bold',pad=10)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='sex', hue='Income',palette='YlGnBu')
plt.title('Income vs Gender',fontweight='bold',pad=10)
plt.xlabel('sex',fontsize=14)
plt.ylabel('count',fontsize=14)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x='Income', y='hours-per-week',palette='YlGnBu')
plt.title('Working Hours per Week vs Income',fontweight='bold',pad=10)
plt.xlabel('income',fontsize=14)
plt.ylabel('hours per week',fontsize=14)
plt.show()
```

```python
plt.figure(figsize=(12,10))
numeric_cols = ['age', 'education-num', 'hours-per-week', 'capital-gain', 'capital-loss', 'Income']
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='YlGnBu', fmt=".2f")
plt.title('Correlation Heatmap (Numeric Features)',fontweight='bold',pad=10)
plt.xticks(rotation=45)
plt.yticks(rotation=45)
plt.show()
```

```python
workclass_cols = [col for col in df.columns if 'workclass_' in col] + ['Income']
plt.figure(figsize=(12,10))
sns.heatmap(df[workclass_cols].corr()[['Income']].sort_values(by='Income', ascending=False),
            annot=True, cmap='YlGnBu', fmt=".2f")
plt.title('Correlation: Workclass Categories vs Income')
plt.show()
```
